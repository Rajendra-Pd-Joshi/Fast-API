```python
from pydantic import BaseModel,Field
from typing import Optional,List
```

```python
class Person(BaseModel):
    weight : float = Field(gt= 0 , le = 120)
    name : str = Field(max_length=20)
    allergies : Optional[List[str]] = Field(default= None,max_length = 20) 
```

# concept of Field validator

### excercise 1 : using field validator validate patient with hdfc and icci employee and give them 20% discount 

```python
from pydantic import BaseModel,Field,field_validator,EmailStr
```

```python
class Patient(BaseModel):
    email : EmailStr

    # writing a custom field validator now
    @field_validator('email')
    @classmethod
    def email_validator(cls,value):
        valid_domains = ['hdfc.com','icici.com']
        #abc@gmail.com
        domain_name = value.split('@')[-1]

        if domain_name not in valid_domains:
            raise ValueError('not a valid hdfc or icici mail')
        return value
```

```python
p1 = Patient(email = 'raju@gmail.com')
```

```python
p2 = Patient(email = 'rajujoshi@hdfc.com')
```

## Operating mode for the field validation
#### mode before type coersion and mode after type coersion

```python
class Person(BaseModel):
    name : str
    age : int # = Field(strict = True) # if this is not commented out it is then type coersion will happen
```

### mode = before  -> the value is passed before type coersion 

```python
class Person(BaseModel):
    name : str
    age : int # = Field(strict = True) # if this is not commented out it is then type coersion will happen

    @field_validator('age',mode= 'before') # the value of age is passed befor type coersion
    @classmethod
    def age_validator(cls,value):
        if type(value) != int:
            raise ValueError('age should be integer')
        return value
```

```python
p1 = Person(name = 'raju',age = '23') # throw and valueerror cause the value passed is before coersion
```

### mode = after  -> the value is passed after type coersion 

```python
class Person(BaseModel):
    name : str
    age : int # = Field(strict = True) # if this is not commented out it is then type coersion will happen

    @field_validator('age',mode= 'after') # the value of age is passed befor type coersion
    @classmethod
    def age_validator(cls,value):
        if type(value) != int:
            raise ValueError('age should be integer')
        return value
```

```python
p1 = Person(name = 'raju',age = '23') # this works fine because age value passed is after coersion i.e 23
```

### problem statement : to register an old patient with age > 60 in their contact details there should be emergency number

### this type of the validation where we need to validate contact_details field based upon another field age > 60 field validator fails. so the solution to this type of the validation is model validtor

## solution for this is Model validator

```python
from pydantic import BaseModel, EmailStr, model_validator
from typing import List, Dict

class Patient(BaseModel):

    name: str
    email: EmailStr
    age: int
    weight: float
    married: bool
    allergies: List[str]
    contact_details: Dict[str, str]

    @model_validator(mode='after')
    def validate_emergency_contact(cls, model):
        if model.age > 60 and 'emergency' not in model.contact_details:
            raise ValueError('Patients older than 60 must have an emergency contact')
        return model
```

```python

def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print('updated')
```

```python

patient_info = {'name':'nitish', 'email':'abc@icici.com', 'age': '65', 'weight': 75.2, 'married': True, 'allergies': ['pollen', 'dust'], 'contact_details':{'phone':'2353462', 'emergency':'235236'}}

patient1 = Patient(**patient_info) 

update_patient_data(patient1)
```

## concept of computed field

```python
from pydantic import BaseModel, EmailStr, computed_field
from typing import List, Dict

class Patient(BaseModel):

    name: str
    email: EmailStr
    age: int
    weight: float # kg
    height: float # mtr
    married: bool
    allergies: List[str]
    contact_details: Dict[str, str]

    @computed_field
    @property
    def calculate_bmi(self)->float:
        bmi = round(self.weight/(self.height**2),2)
        return bmi
```

```python
def update_patient_data(patient: Patient):

    print(patient.name)
    print(patient.age)
    print(patient.allergies)
    print(patient.married)
    print('BMI', patient.calculate_bmi)
    print('updated')
```

```python
patient_info = {'name':'nitish', 'email':'abc@icici.com', 'age': '65', 'weight': 75.2, 'height': 1.72, 'married': True, 'allergies': ['pollen', 'dust'], 'contact_details':{'phone':'2353462', 'emergency':'235236'}}

patient1 = Patient(**patient_info) 
```

```python
update_patient_data(patient1)
```

## concept of serialization

```python
from pydantic import BaseModel

class Address(BaseModel):

    city: str
    state: str
    pin: str

class Patient(BaseModel):

    name: str
    gender: str = 'Male'
    age: int
    address: Address

address_dict = {'city': 'gurgaon', 'state': 'haryana', 'pin': '122001'}

address1 = Address(**address_dict)

patient_dict = {'name': 'nitish', 'age': 35, 'address': address1}

patient1 = Patient(**patient_dict)

temp = patient1.model_dump(exclude_unset=True)

print(temp)
print(type(temp))
```
